This visualization of the data does not really give us any idea of the possible correlation, there is just too much going on. But if i look at just one year:

```python
df2020 = df.loc[14975:, ['date', 'mean_temp', 'precipitation', 'pressure']]
df20202 = df.loc[14975:, ['mean_temp', 'precipitation', 'pressure']]
df2020
```

```python
plt.plot(df2020['mean_temp'], 'r')
plt.title('Mean Temperature')
plt.show()
```

```python
plt.plot(df2020['precipitation'], 'r')
plt.title('Precipitation')
plt.show()
```

```python
plt.plot(df2020['pressure'], 'r')
plt.title('Pressure')
plt.show()
```

Just by looking at the graphs I can't see any obvious correlation between the mean temperature and the others, but by looking at the precipitation and pressure i can see that when pressure is high, precipitation seems lower. If you look at early in the year, pressure is quite high, and precipitation is lower than later in the year when pressure drops. Because of this I suspect some degree of negative correlation between these two. 

```python
df20202.corr()
```

As you can see there is a slight negative correlation between pressure and precipitation which matches my obersvation above. I will now show the correlation through all the years to see whether this was just a fluke in 2020.

```python
df.loc[:, ['mean_temp', 'precipitation', 'pressure']].corr()
```

The correlation is still there even when looking at every year which tells me that there most likely is some causality, the correlation is weaker here though compared to looking at just 2020. Now I would like to see if there is any correlation between any of the other attributes.
